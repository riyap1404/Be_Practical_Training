Python follows an order of operations very similar to BODMAS/PEMDAS, evaluated through strict operator precedence rules.

Parentheses (): Evaluated first. You can use nested or standard brackets to override any other precedence.


Exponentiation **: Evaluated second (equivalent to the 'O' / 'Orders' or exponents in BODMAS/PEMDAS).

Multiplication, Division, Floor Division, and Modulo *, /, //, %: Evaluated third. These operators share the same precedence level and execute from left to right.

Addition and Subtraction +, -: Evaluated last. They also share an equal precedence level and execute from left to right.

```python
#activity
fruits = ["apple", "banana","mango","orange"]
print(fruits)
print(len(fruits))

fruits.append("grapes")
more_fruits= ["pineapple","cherry"]
fruits.extend(more_fruits)
print("after extend:", fruits)

print("first fruits:", fruits[0])
print("last fruit:", fruits[-1])
print(fruits[2:5])

fruits[1] = "strawberry"
print("after updating index 1:", fruits)
fruits.remove("mango")
print("after removing mango  ", fruits)

removed_fruit= fruits.pop(2)
print(removed_fruit)
print("after removed fruits", fruits)

fruits.sort()
print("sort:", fruits)
fruits.reverse()
print("reverse : ", fruits)

fruits.clear()
print(fruits)
```
